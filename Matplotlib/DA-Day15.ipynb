```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
 # plt.figure(figsize=(h,w))
# It is used to adjust the height and width of the chart
```

```python
x=np.arange(1,11)
y=x**2
z=x**3
```

```python
plt.figure(figsize=(15,5))
plt.plot(x,y,label='x-axis',linewidth=2,marker='o', markersize=5, markerfacecolor='green',markeredgecolor='black')
plt.plot(x,z,linestyle='dotted',label='z-axis',linewidth=2,marker='o', markersize=5,color='red', markerfacecolor='green',markeredgecolor='black')
plt.legend(title='Values',fontsize=30,loc='upper center')
plt.xlabel("value to 1 to 10",fontsize=16)
plt.ylabel("square of 1 to 10",fontsize=16)
plt.title('Graph',fontsize=16,color='blue')


# Annotate 
for i in range(len(x)):
    plt.annotate(f'({x[i]},{y[i]})',(x[i],y[i]),va='top',ha='left')

for i in range(len(z)):
    plt.annotate(f'({x[i]},{z[i]})',(x[i],z[i]),va='top',ha='left')

plt.grid()
```
