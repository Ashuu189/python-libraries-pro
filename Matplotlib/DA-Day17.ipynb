```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales=pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/sales_data_boolean_filtering.csv")
```

```python
# 3rd sub-plot():
# syntax 
        #  var,axes=plt.subplot(nrow=value,ncol=value)
        #  axes[row,col].plot()
x=np.arange(1,11)
y=x**2
```

```python
rg_gp=sales.groupby('Region')['TotalSales'].sum()
rg_gp=pd.DataFrame(rg_gp)
rg_gp.reset_index(inplace=True)

plt.plot(rg_gp['Region'],rg_gp['TotalSales'])
```

```python
fig,axes=plt.subplots(nrows=2,ncols=2)
axes[0,0].plot(x,y)
axes[0,0].set_facecolor('green')
axes[0,0].set_xlabel('values')
axes[0,0].set_ylabel('square')
axes[0,1].plot(rg_gp['Region'],rg_gp['TotalSales'])
```

```python
fig,axes=plt.subplots(nrows=3,ncols=3)
axes[0,0].plot(x,y)
axes[0,2].plot(x,y)
axes[1,1].plot(x,y)
axes[2,0].plot(x,y)
axes[2,2].plot(x,y)
plt.tight_layout()
```
