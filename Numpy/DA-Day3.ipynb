```python
ages=np.array([[17,34,65,23,72,18,45,12,25,11],
              [22,42,16,46,38,12,62,54,15,18]])

```

```python
# Find out the ages greater than 60
ages[ages> 18]
```

```python
# Logical operator
# And(&)
# OR(|)
# NOT(~)

```

```python
ages[(ages> 18) & (ages< 60)]
```

```python
ages[(ages%5==0) | (ages%7==0)]
```

```python
ages[(ages>18) | (ages>60)]  
```
