```python
# Concatenation of DataFrames
# It is a process of merging two or more DataFrames,
# Syntax:
# 1) Vertical-> row by row
        # pd.concat([var1,var2...var n])
# Horizontal-> column by column
        # pd.concat([var1,var2...var n],axis=1)
```

```python
d={'ENAME':['SMITH','ALLEN','WARD','JONES','MARTIN'],
   'Eid':  [,7499,7521,7566,7698],
   'Salary':['8LPA','10LPA','9LPA','12LPA','15LPA']}
```

```python
df=pd.DataFrame(d)
```

```python
df
```

```python
e={'STUDENT_ID':['01','02','03','04','05'],
   'STUDENT_NAME':['SMITH','ALLEN','WARD','JONES','MARTIN'],
   'STUDENT_COURSE':['BTECH','BSC','BA','LLB','MTECH']}
```

```python
ef
```

```python
pd.concat([df,ef])
```

```python
# # Joins:
# When we want to fetch the data from 2 different
# DataFrames simultaneously we use joins.

# Types:
# 1) Right()
# 2) Left()-> 
# 3) inner()
# 4) outer()

# Syntax-> 
# pd.merge(var1,var2,on='common_col',how='inner/outer/left/right')
```

```python
df_1=pd.DataFrame({'Blood_Group':['A-','B+','O-','A+'],
                   'Eid':[105,104,103,106],
                   'Bonus':['12K','3K','5K','9K']})
```

```python
df_1
```

```python
d={'ENAME':['SMITH','ALLEN','WARD','JONES'],
   'Eid':  [105,104,103,106],
   'Salary':['8LPA','10LPA','9LPA','12LPA']}
```

```python
df=pd.DataFrame(d)
```

```python
pd.merge(df,df_1,on='Eid',how='inner')
```
