```python
import pandas as pd
import numpy as np
```

```python
titanic=pd.read_csv('https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/titanic.csv'
'')
```

```python
titanic['surname']=titanic['name'].str.split(',').str[0]
```

```python
titanic.head()
```

```python
titanic['prefix']=titanic['name'].str.split('.').str[0]
```

```python
titles = []

for i in titanic['prefix']:
    parts = str(i).split(',')

    if len(parts) > 1:
        titles.append(parts[1].strip())
    else:
        titles.append(parts[0].strip())

titanic['prefix'] = titles
        
                                
```
