```python
import pandas as pd
import numpy as np
```

```python
flat=pd.read_csv('https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Cleaning/flats_uncleaned.csv')
```

```python
flat.drop_duplicates(inplace=True)
```

```python
flat['price'].unique()
```

```python
out=[]
for i in flat['price']:
    if 'Lac' in i:
        i=i.strip('₹ Lac')
        f=float(i)
        out.append(f*100000)
    elif 'Cr' in i:
        l=i.strip('₹ Cr')
        u=float(l)
        out.append(u*10000000)
    else:
        np.nan

out
```
