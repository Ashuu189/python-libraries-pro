```python
# Loc and iloc
# loc---> label based indexing:
# syntax:
# var.Loc[index]
# var.Loc[[index1,index2,..........indexn]]
# var.loc[SV,EV,UP]

#iloc:location bases indexing
# var.iLoc[index]
# var.iLoc[[index1,index2,..........indexn]]
# var.iloc[SV,EV,UP]
```

```python
s.loc['messi']
```

```python
s.loc[['messi','pedri','rodri','scala']]
```
