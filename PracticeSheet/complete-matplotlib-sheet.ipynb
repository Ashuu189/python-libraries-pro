```python
sales_avg=hr[hr['department']=='sales']['average_montly_hours'].mean()
accounting_avg=hr[hr['department']=='accounting']['average_montly_hours'].mean()
hr_avg=hr[hr['department']=='hr']['average_montly_hours'].mean()
technical_avg=hr[hr['department']=='technical']['average_montly_hours'].mean()
support_avg=hr[hr['department']=='support']['average_montly_hours'].mean()
management_avg=hr[hr['department']=='management']['average_montly_hours'].mean()
IT_avg=hr[hr['department']=='IT']['average_montly_hours'].mean()
product_mng_avg=hr[hr['department']=='product_mng']['average_montly_hours'].mean()
marketing_avg=hr[hr['department']=='marketing']['average_montly_hours'].mean()
RandD_avg=hr[hr['department']=='RandD']['average_montly_hours'].mean()


avg_hours = [
    sales_avg,
    accounting_avg,
    hr_avg,
    technical_avg,
    support_avg,
    management_avg,
    IT_avg,
    product_mng_avg,
    marketing_avg,
    RandD_avg
]

departments = [
    'sales',
    'accounting',
    'hr',
    'technical',
    'support',
    'management',
    'IT',
    'product_mng',
    'marketing',
    'RandD'
]

```
