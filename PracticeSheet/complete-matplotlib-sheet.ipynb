```python
import matplotlib.pyplot as plt

# 1. Define data
categories = ['Men', 'Women', 'Children']
values = [45, 35, 20]
colors = ['#ff7f0e', '#1f77b4', '#2ca02c']  # Orange, Blue, Green

# 2. Format labels with counts and percentages
total = sum(values)
legend_labels = [f'{cat}: {val} ({val/total*100:.1f}%)' for cat, val in zip(categories, values)]
table_data = [[cat, val, f'{val/total*100:.1f}%'] for cat, val in zip(categories, values)]

# 3. Create figure and pie plot
fig, ax = plt.subplots(figsize=(8, 7))

wedges, texts, autotexts = ax.pie(
    values,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.75
)

# 4. Add side legend mapping colors to categories and values
ax.legend(
    wedges,
    legend_labels,
    title="Demographics",
    loc="center left",
    bbox_to_anchor=(1.05, 0.7)
)

# 5. Add matching data table beneath the pie chart
table = ax.table(
    cellText=table_data,
    colLabels=['Category', 'Count', 'Share'],
    rowColours=colors,
    cellLoc='center',
    loc='bottom',
    bbox=[0.1, -0.35, 0.8, 0.25]  # [left, bottom, width, height]
)

# Style table text size
table.auto_set_font_size(False)
table.set_fontsize(10)

plt.title('Category Distribution with Color-Mapped Table & Legend', pad=20)
plt.subplots_adjust(bottom=0.3, right=0.75)
plt.show()
```
