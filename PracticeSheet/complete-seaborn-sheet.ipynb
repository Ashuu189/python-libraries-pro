```python
sns.boxplot(x='quality',data=wine_quality)
```

```python
q1=wine_quality['quality'].quantile(0.25)
q3=wine_quality['quality'].quantile(0.75)

q1,q3
```

```python
iqr=q3-q1
iqr
```

```python
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr

lower_bound,upper_bound
```

```python
outlier=wine_quality[(wine_quality['quality'] < lower_bound) | (wine_quality['quality']>upper_bound)].count()
```

```python
outlier
```

```python
out=[]
for i in wine_quality['quality']:
    if i > upper_bound:
        out.append(upper_bound)
    elif i<lower_bound:
        out.append(lower_bound)
    else:
        out.append(i)

wine_quality['quality']=out
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker='o')

plt.annotate(
    'Highest',
    xy=(4, 30),
    xytext=(3, 35),
    arrowprops=dict(arrowstyle='->')
)

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker='o')

for i in range(len(x)):
    plt.annotate(
        f'({x[i]}, {y[i]})',
        (x[i], y[i]),
        xytext=(5, 5),
        textcoords='offset points'
    )

plt.show()
```
