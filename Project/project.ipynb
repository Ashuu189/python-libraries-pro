```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales=pd.read_csv("C:\\Users\\ashut\\Downloads\\sales_project.csv", encoding='ISO-8859-1')

```

```python
sales.head(5)
```

```python
sales=sales.drop(['Status','unnamed1'], axis=1)
```

```python
sales.head()
```

```python
sales.duplicated().sum()
```

```python
sales=sales.drop_duplicates()
```

```python
sales.duplicated().sum()
```

```python
sales.info()
```

```python
sales['Amount'].isna().sum()

sales.dropna(inplace=True)

sales['Amount'].isna().sum()
```

```python
sales.head()
```

```python
d={'F':'Female','M':'Male'}

sales['Gender']=sales['Gender'].map(d)
```

```python
sales['Gender'].values
```

```python
sales['Marital_Status']=sales['Marital_Status'].astype('str')

sales['Marital_Status'].info
```

```python
d={'0':'Single','1':'Married'}
sales['Marital_Status']=sales['Marital_Status'].map(d)
```

```python
sales['Marital_Status'].values
```

```python
sales.head()
```

```python
sales['State'].unique()
```

```python
sales['Occupation'].unique()
```

```python
def label(a):
    for i in a.containers:
        a.bar_label(i, label_type='center', fontsize=20, color='black', padding=3)
```

```python
fig,axes= plt.subplots(nrows=1,ncols=2,figsize=(20,10))
c_g=sns.countplot(x='Gender',data=sales,ax=axes[0],hue='Gender')
label(c_g)
```

```python
ge_gender=sales.groupby('Gender')['Amount'].sum()
ge_gender=pd.DataFrame(ge_gender)
ge_gender.reset_index(inplace=True)
a=plt.pie(ge_gender['Amount'],labels=ge_gender['Gender'].unique(),autopct='%0.0f%%',colors=['#ff9999','#66b3ff'])

plt.show()
```

```python
ge_gender=sales.groupby('Gender')['Amount'].sum()
ge_gender=pd.DataFrame(ge_gender)
ge_gender.reset_index(inplace=True)
axes[1].pie(ge_gender['Amount'],labels=ge_gender['Gender'].unique(),autopct='%0.0f%%',colors=['#ff9999','#66b3ff'])

plt.show()
```

```python
fig, axes = plt.subplots(3, 2, figsize=(18, 18))

# ---------------- AGE GROUP ----------------
age_g = sales.groupby('Age Group')['Amount'].sum().reset_index()

sns.barplot(
    x='Age Group',
    y='Amount',
    data=age_g,
    hue='Age Group',
    ax=axes[0, 0],
    legend=False
)

axes[0, 0].set_title('Amount by Age Group')
axes[0, 0].set_xlabel('Age Group')
axes[0, 0].set_ylabel('Amount')

for container in axes[0, 0].containers:
    axes[0, 0].bar_label(container)


# ---------------- STATE ----------------
state_gr = sales.groupby('State')['Amount'].sum().reset_index()

sns.barplot(
    x='State',
    y='Amount',
    data=state_gr,
    hue='State',
    palette='coolwarm',
    ax=axes[0, 1],
    legend=False
)

axes[0, 1].set_title('Amount by State')
axes[0, 1].set_xlabel('State')
axes[0, 1].set_ylabel('Amount')

axes[0, 1].tick_params(axis='x', rotation=45)

for container in axes[0, 1].containers:
    axes[0, 1].bar_label(container)


# ---------------- OCCUPATION ----------------
occupation_gr = sales.groupby('Occupation')['Amount'].sum().reset_index()

sns.barplot(
    x='Occupation',
    y='Amount',
    data=occupation_gr,
    hue='Occupation',
    palette='viridis',
    errorbar=None,
    estimator=sum,
    ax=axes[1, 0],
    legend=False
)

axes[1, 0].set_title('Amount by Occupation')
axes[1, 0].set_xlabel('Occupation')
axes[1, 0].set_ylabel('Amount')

axes[1, 0].tick_params(axis='x', rotation=45)

for container in axes[1, 0].containers:
    axes[1, 0].bar_label(container)


# ---------------- PRODUCT CATEGORY ----------------
product_gr = sales.groupby('Product_Category')['Amount'].sum().reset_index()

sns.barplot(
    x='Product_Category',
    y='Amount',
    data=product_gr,
    hue='Product_Category',
    palette='viridis',
    errorbar=None,
    estimator=sum,
    ax=axes[1, 1],
    legend=False
)

axes[1, 1].set_title('Amount by Product Category')
axes[1, 1].set_xlabel('Products')
axes[1, 1].set_ylabel('Amount')

axes[1, 1].tick_params(axis='x', rotation=45)

for container in axes[1, 1].containers:
    axes[1, 1].bar_label(container)


# ---------------- MARITAL STATUS ----------------
marry_gr = sales.groupby('Marital_Status')['Amount'].sum().reset_index()

sns.barplot(
    x='Marital_Status',
    y='Amount',
    data=marry_gr,
    hue='Marital_Status',
    palette='viridis',
    errorbar=None,
    estimator=sum,
    ax=axes[2, 0],
    legend=False
)

axes[2, 0].set_title('Amount by Marital Status')
axes[2, 0].set_xlabel('Marital Status')
axes[2, 0].set_ylabel('Amount')

for container in axes[2, 0].containers:
    axes[2, 0].bar_label(container)


# Remove empty subplot
fig.delaxes(axes[2, 1])

# Automatically adjust spacing
plt.tight_layout()

plt.show()
```
