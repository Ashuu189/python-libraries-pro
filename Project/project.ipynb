```python
def label(a):
    for i in a.containers:
        a.bar_label(i, label_type='center', fontsize=20, color='black', padding=3)
```

```python
fig,axes= plt.subplots(nrows=1,ncols=2,figsize=(20,10))
c_g=sns.countplot(x='Gender',data=sales,ax=axes[0],hue='Gender')
label(c_g)
```

```python
ge_gender=sales.groupby('Gender')['Amount'].sum()
ge_gender=pd.DataFrame(ge_gender)
ge_gender.reset_index(inplace=True)
a=plt.pie(ge_gender['Amount'],labels=ge_gender['Gender'].unique(),autopct='%0.0f%%',colors=['#ff9999','#66b3ff'])

plt.show()
```
