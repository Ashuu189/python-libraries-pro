```python
# # Distribution plot:-
# # It helps us to understand how the data got distributed.

# # Distribution is spread of value.

# # By this we can analyse min, max, highly dense values.
# # Unique values

# # Types of distribution:

# # 1) histplot
# # 2) joinplot
# # 3) pairplot()

# 1) histplot-> It is used to understand the distribution of the data based on 1-column.

# Syntax: 
# sns.histplot(var['col name'])

#             or 

# sns.histplot(x='col_name', data=var)


# palette= set1,set2,set3, inferno, magma, viridis, reds, greens, purples, coolwarm


```

```python
import pandas as pd
import seaborn as sns
```

```python
hr=pd.read_csv('https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/hr_data.csv')
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12,7))
a=sns.histplot(x='department', data=hr, hue='salary', palette='inferno')
plt.xticks(rotation=45)
for i in a.containers:
    a.bar_label(i)
plt.show()
```

```python
tips=pd.read_csv('https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/tips.csv')
```

```python
tips.head()
```

```python
a=sns.histplot(x='smoker', data=tips, hue='sex', palette='magma')

for i in a.containers:
    a.bar_label(i)
```

```python
# Joint Plot: 
# It is used to understand the distribution of the data based on two columns.

# Syntax:
# sns.jointplot(x='col_name',y='col_name', data=var)
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips, hue='sex')
```

```python
# It is a combination of scatter and hist plot
```
