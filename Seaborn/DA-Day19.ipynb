```python
# IQR=
# q3=var['col_name'].quantile(0.75)
# q1=var['col_name'].quantile(0.25)

```

```python
q3=titanic['age'].quantile(0.75)
q1=titanic['age'].quantile(0.25)
```

```python
q3,q1
```

```python
iqr=q3-q1
```

```python
iqr
```

```python
lower_bound=q1-0.5*iqr
upper_bound=q3+0.5*iqr


lower_bound,upper_bound
```

```python
outlier=titanic[(titanic['age']< lower_bound) | (titanic['age']> upper_bound)].count()
```

```python
outlier
```

```python
# How to handle outliers

out=[]
for i in titanic['age']:
    if i>upper_bound:
        out.append(upper_bound)
    elif i<lower_bound:
        out.append(lower_bound)
    else:
        out.append(i)

titanic['age']=out
```
