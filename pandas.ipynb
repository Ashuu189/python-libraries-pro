```python
appliances=appliances.drop(['goods-title-link--jump href','goods-title-link--jump','goods-title-link'],axis=1)
```

```python
appliances.head()
```

```python
appliances['price']=appliances['price'].str.strip('$')

appliances['price']=appliances['price'].str.replace(',','')

appliances['price'].unique()

appliances['price']=appliances['price'].astype(float)
    
```

```python
price=[]
for i in appliances['price']:
    output= 'Rs'+str(i*95)
    price.append(output)

appliances['price_in_inr']=price
```
